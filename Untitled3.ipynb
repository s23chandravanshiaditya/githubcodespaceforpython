```python
a=int(input("enter the first number:"))
b=int(input("enter the second number:"))
operator=input("enter the operator:")
if operator == '*':
  c=a*b
  print(c)
elif operator=='+':
  c=a+b
  print(c)
elif operator=='-':
  c=a-b
  print(c)
elif operator=='/':
  if b==0:
    print("invalid division")
  else:
    c=a/b
    print(c)
else:
  print("invalid operation")
```
